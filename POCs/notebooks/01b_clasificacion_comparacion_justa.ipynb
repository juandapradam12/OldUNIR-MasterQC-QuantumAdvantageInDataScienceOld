## Two tables (do not mix them)

- **Table A — kernel-fair:** same rows, same features (ZZ-ready `n_qubits` angles in `[-π, π]`), same split. Isolates ZZ fidelity vs RBF / linear / GB / logistic.
- **Table B — matched n, full features:** same rows and split; classical models keep every column; QSVM still uses PCA to `n_qubits`. This answers “should you replace GB on this table?”, not “is the kernel better than RBF”.

Preprocess is fit on **train only**. Quantum-native data is not re-scaled (it is already ZZ angles).
